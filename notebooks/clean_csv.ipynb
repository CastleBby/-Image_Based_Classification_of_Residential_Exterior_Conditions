Due to mismatch in images and csv label rows, need to create a clean csv label file 

```python
import pandas as pd

df = pd.read_csv("../data/house_labels.csv")

df = df[df["image_number"].notna()]

def to_filename(x):
    x = str(x)

    if x.startswith("img_"):
        return f"{x}.png" if not x.endswith(".png") else x

    try:
        return f"img_{int(float(x)):03d}.png"
    except:
        return None

df["image_name"] = df["image_number"].apply(to_filename)

# drop rows that failed conversion
df = df[df["image_name"].notna()]

cols_to_keep = ["image_name", "label"]
extra_cols = [col for col in df.columns if col not in ["image_number", "image_name", "label"]]
df = df[cols_to_keep + extra_cols]

df.to_csv("../data/labels_fixed.csv", index=False)

print(f"Created labels_fixed.csv with {len(df)} rows")
```

```python
import os
import pandas as pd

image_dir = "../data/house_images"

# collect real images
real_images = set()
for cls in os.listdir(image_dir):
    class_path = os.path.join(image_dir, cls)
    if os.path.isdir(class_path):
        for img in os.listdir(class_path):
            if img.endswith(".png"):
                real_images.add(img)

df = pd.read_csv("data/labels_fixed.csv")

df = df[df["image_name"].isin(real_images)]

df.to_csv("data/labels_clean.csv", index=False)

print(f"Final rows: {len(df)}")
```
